First, we'll import the `Image` module from `PIL`. Then, we'll define a function `resize_image` that takes the image path and a percentage as input. This function will open the image, calculate new dimensions, resize it, and save the resized image with a new name. Finally, we'll demonstrate its usage by creating a dummy image and resizing it.

```python
from PIL import Image
from IPython.display import Image as DisplayImage

def resize_image(image_path, percentage):
    """
    Resizes an image to a given percentage of its original width and height.

    Args:
        image_path (str): The path to the input image file.
        percentage (float): The percentage (e.g., 50 for 50%) to resize the image.

    Returns:
        str: The path to the resized image file.
    """
    try:
        with Image.open(image_path) as img:
            original_width, original_height = img.size
            new_width = int(original_width * (percentage / 100))
            new_height = int(original_height * (percentage / 100))

            resized_img = img.resize((new_width, new_height))

            # Save the resized image with a new name
            file_name, file_extension = image_path.rsplit('.', 1)
            output_path = f"{file_name}_resized_{percentage}.{file_extension}"
            resized_img.save(output_path)
            print(f"Image resized and saved to: {output_path}")
            return output_path
    except FileNotFoundError:
        print(f"Error: Image file not found at {image_path}")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None

# --- Demo Usage with user-provided image ---

user_image_path = '/content/HD-wallpaper-jr-ntr-smile-ntr.jpg'

# Resize the user's image to 50% of its original size
resized_image_path = resize_image(user_image_path, 50)

# Display the original and resized images in Colab
print("\nOriginal Image:")
display(DisplayImage(filename=user_image_path))

if resized_image_path:
    print("\nResized Image:")
    display(DisplayImage(filename=resized_image_path))
```
