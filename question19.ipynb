```python
# Write a program to create a `256 × 256` grayscale image in which every pixel has an intensity value of `128`.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a 256 × 256 grayscale image
# Every pixel has an intensity value of 128
image = np.full((256, 256), 128, dtype=np.uint8)

# Display the image
plt.imshow(image, cmap="gray", vmin=0, vmax=255)
plt.title("256 × 256 Grayscale Image - Intensity 128")
plt.axis("off")
plt.show()

# Print image details
print("Image size:", image.shape)
print("Pixel intensity value:", image[0, 0])
```

# **Conclusion**

The program successfully creates and displays a 256 × 256 grayscale image with an intensity value of 128 for every pixel. Since all pixels have the same intensity, the entire image appears as a uniform gray shade.
