```python
# Write a program to create and display a grayscale intensity ramp whose intensity gradually changes from 0 to 255.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a 256 × 256 grayscale intensity ramp
ramp = np.tile(np.arange(256, dtype=np.uint8), (256, 1))

# Display the intensity ramp
plt.imshow(ramp, cmap="gray", vmin=0, vmax=255)
plt.title("Grayscale Intensity Ramp (0 to 255)")
plt.xlabel("Intensity: 0 → 255")
plt.axis("off")
plt.show()
```

# **Conclusion**

**The program successfully creates and displays a grayscale intensity ramp where pixel intensity gradually increases from 0 to 255, demonstrating the complete range of grayscale intensities.**
