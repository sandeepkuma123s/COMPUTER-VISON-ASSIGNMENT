```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Create a dummy image for demonstration purposes if no image is uploaded
try:
    # Try to load an image, assuming it's uploaded to '/content/image.jpg'
    # You can change 'image.jpg' to the actual path of your image
    img = cv2.imread('/content/MAHIBHAI.JPG')
    if img is None:
        raise FileNotFoundError
except FileNotFoundError:
    print("No image found at '/content/MAHIBHAI.JPG'. Creating a dummy image for demonstration.")
    # Create a dummy image if no file is found
    img = np.zeros((300, 500, 3), dtype=np.uint8) # Height, Width, Channels
    img = cv2.putText(img, 'Dummy Image', (100, 150), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2, cv2.LINE_AA)

# Get the dimensions of the image
# The shape attribute returns (height, width, channels)
height, width, channels = img.shape

# Print the dimensions
print(f"Image Height: {height} pixels")
print(f"Image Width: {width} pixels")
print(f"Number of Channels: {channels}")

# Display the image (optional)
print("\nDisplaying the image:")
cv2_imshow(img)
```
