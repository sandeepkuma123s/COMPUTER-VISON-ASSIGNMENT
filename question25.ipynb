```python
# Write a program to rotate an image by 90 degrees, display it, and save the rotated image.
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get filename
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

# Check whether image was loaded
if img is None:
    print("Error: Image could not be loaded.")
else:
    # Rotate image by 90 degrees clockwise
    rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # Save rotated image
    cv2.imwrite("/content/lotus2.jpg", rotated_img)

    # Convert BGR to RGB for Matplotlib
    rotated_rgb = cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB)

    # Display rotated image
    plt.imshow(rotated_rgb)
    plt.axis("off")
    plt.show()

    print("Image rotated by 90 degrees and saved successfully.")
```
