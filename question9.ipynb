```python
# Write a program to display an image using Matplotlib and hide the axis.
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get filename
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

# Check if image was loaded
if img is None:
    print("Error: Image could not be loaded.")
else:
    # Convert BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display image
    plt.imshow(img_rgb)

    # Hide axis
    plt.axis("off")

    # Show image
    plt.show()
```

1.  **Import Matplotlib**: We import the `matplotlib.pyplot` module, commonly aliased as `plt`.
2.  **Display Image**: `plt.imshow(color_image_rgb)` is used to display the `color_image_rgb` array as an image. This array is a NumPy array representing the image data.
3.  **Set Title**: `plt.title()` adds a descriptive title to the image plot.
4.  **Hide Axes**: `plt.axis('off')` is called to turn off the axis lines and labels, resulting in a cleaner image display.
5.  **Show Plot**: `plt.show()` renders the image plot.
