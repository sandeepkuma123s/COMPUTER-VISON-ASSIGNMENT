```python
# Write a program to calculate and print the minimum and maximum intensity values of a grayscale image.
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image in grayscale
image = cv2.imread("eagle.jpg", cv2.IMREAD_GRAYSCALE)

# Check whether image is loaded successfully
if image is None:
    print("Error: Image could not be loaded.")
else:
    # Calculate minimum and maximum intensity values
    min_intensity = image.min()
    max_intensity = image.max()

    # Print the values
    print("Minimum intensity value:", min_intensity)
    print("Maximum intensity value:", max_intensity)

    # Display the grayscale image
    plt.imshow(image, cmap="gray")
    plt.title(f"Grayscale Image\nMin: {min_intensity}, Max: {max_intensity}")
    plt.axis("off")
    plt.show()
```

# **Conclusion**

The program successfully reads a grayscale image and calculates its minimum and maximum intensity values. It also displays the grayscale image along with these values. This helps in understanding the darkest and brightest pixels present in the image and is useful in basic image processing and enhancement.
