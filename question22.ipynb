```python
# Write a program to convert an 8-bit grayscale image into a 2-bit quantized image and display the result.
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("/content/thar.webp", cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image could not be loaded.")
else:
    # Convert 8-bit image to 2-bit
    quantized = image // 64

    # Convert back to displayable grayscale values
    display_image = quantized * 85

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image, cmap="gray", vmin=0, vmax=255)
    plt.title("Original 8-bit Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(display_image, cmap="gray", vmin=0, vmax=255)
    plt.title("2-bit Quantized Image")
    plt.axis("off")

    plt.show()

    print("Original intensity levels: 256")
    print("Quantized intensity levels: 4")
```


# Explanation

An 8-bit grayscale image has 256 intensity levels (0–255). A 2-bit image has only 4 intensity levels (0–3).

The 256 intensity values are divided into 4 groups, and each group is represented by one intensity level. This reduces the number of gray shades and produces a 2-bit quantized image.

# Output
Original image: 256 intensity levels
Quantized image: 4 intensity levels
The result has fewer shades of gray and reduced image detail.
# Conclusion

The 8-bit grayscale image is successfully converted into a 2-bit quantized image with 4 intensity levels.
