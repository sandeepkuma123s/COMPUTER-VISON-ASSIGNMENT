```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the image
img = cv2.imread("/content/MAHIBHAI2.JPG")

# Check if the image was loaded successfully
if img is None:
    print("Warning: 'image.jpg' not found. Displaying a placeholder image instead.")
    # Create a black placeholder image (e.g., 200x200 pixels)
    img = np.zeros((200, 200, 3), dtype=np.uint8)
    # Optionally, add some text to the placeholder
    cv2.putText(img, "No Image Found", (10, 100), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)

# Display the image
cv2_imshow(img)

# Wait until a key is pressed (This will not work as expected in Colab for cv2_imshow)
# cv2.waitKey(0)

# Close all windows (This is not necessary with cv2_imshow as it displays directly in the notebook)
# cv2.destroyAllWindows()
```
