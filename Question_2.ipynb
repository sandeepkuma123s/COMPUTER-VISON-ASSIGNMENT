```python
import cv2
from google.colab import files
```

Please upload an image file when prompted. For example, you can upload `lotus21.jpg` if you have it, or any other image.

```python
uploaded = files.upload()

# Get the filename of the uploaded file
if uploaded:
    filename = next(iter(uploaded))
    print(f"Uploaded file: {filename}")
else:
    filename = None
    print("No file was uploaded.")
```

```python
if filename:
    # Read the image using OpenCV
    img = cv2.imread(filename)

    # Check if the image was loaded successfully
    if img is None:
        print(f"Error: Could not load image '{filename}'. Please check the filename and ensure it's a valid image file.")
    else:
        print(f"Image '{filename}' loaded successfully. Image dimensions: {img.shape}")
else:
    print("Cannot check image loading as no file was uploaded.")
```

```python
import matplotlib.pyplot as plt

# Display the loaded image
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title(f"Loaded Image: {filename}")
plt.axis('off') # Hide axes
plt.show()
```
