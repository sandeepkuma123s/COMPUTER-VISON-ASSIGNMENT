Now, let's import the necessary libraries and download a sample image to work with. We'll use `cv2` for image processing and `matplotlib.pyplot` for displaying the image.

```python
import cv2
import matplotlib.pyplot as plt
import urllib.request

# Download a sample image
# Using a different, more reliable public domain image URL from GitHub
image_url = 'https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg'
original_filename = 'lena.jpg'
urllib.request.urlretrieve(image_url, original_filename)

print(f"Downloaded '{original_filename}'")
```

Now, we will read the downloaded image, display it, and then save it with a new filename. When reading an image with `cv2.imread`, it's often read in BGR format, so we convert it to RGB for correct display with `matplotlib`.

```python
# Read the image
img = cv2.imread(original_filename)

# Check if the image was loaded successfully
if img is None:
    print(f"Error: Could not load image '{original_filename}'")
else:
    # Display the original image using matplotlib
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title('Original Image')
    plt.axis('off') # Hide axes
    plt.show()

    # Define the new filename
    new_filename = 'lena_copy.jpg'

    # Save the image with a different filename
    cv2.imwrite(new_filename, img)

    print(f"Image saved as '{new_filename}'")
```
