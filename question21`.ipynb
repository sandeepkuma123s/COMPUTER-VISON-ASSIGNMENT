```python
# Write a program to convert an 8-bit grayscale image into a 4-bit quantized image and display the result.
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the 8-bit grayscale image
image = cv2.imread("/content/thar.webp", cv2.IMREAD_GRAYSCALE)

# Check whether image is loaded successfully
if image is None:
    print("Error: Image could not be loaded.")
else:
    # Convert 8-bit image (0-255) to 4-bit (0-15)
    quantized = image // 16

    # Convert 4-bit values back to 8-bit range for display
    display_image = quantized * 16

    # Display original and quantized images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image, cmap="gray", vmin=0, vmax=255)
    plt.title("Original 8-bit Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(display_image, cmap="gray", vmin=0, vmax=255)
    plt.title("4-bit Quantized Image")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # Print information
    print("Original image intensity levels: 256")
    print("Quantized image intensity levels: 16")
```

# **Explanation**

An 8-bit grayscale image contains 256 intensity levels (0–255). In 4-bit quantization, these levels are reduced to 16 intensity levels (0–15).

The image is divided into intensity ranges, and similar pixel values are grouped together. As a result, the quantized image contains fewer shades of gray and may appear slightly less detailed.

### Conclusion: The program reduces the image from 256 intensity levels to 16 levels and displays both the original and quantized images.
