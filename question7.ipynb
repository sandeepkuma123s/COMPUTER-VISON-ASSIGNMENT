```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Create a dummy image for demonstration if no image is available
# In a real scenario, replace 'sample_image.jpg' with your image file path
try:
    # Attempt to read a sample image (replace with your actual image path)
    image_path = '/content/dog.webp'
    # Download a sample image if not available (for demonstration purposes)
    import requests
    import os

    if not os.path.exists(image_path):
        print(f"Downloading a sample image to {image_path}...")
        image_url = 'https://upload.wikimedia.org/wikipedia/commons/thumb/e/ec/Mona_Lisa%2C_by_Leonardo_da_Vinci%2C_from_C2RMF_retouched.jpg/400px-Mona_Lisa%2C_by_Leonardo_da_Vinci%2C_from_C2RMF_retouched.jpg'
        response = requests.get(image_url)
        response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
        with open(image_path, 'wb') as f:
            f.write(response.content)
        print("Download complete.")

    # Read the image directly in grayscale mode
    gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if gray_image is None:
        raise FileNotFoundError(f"Image not found at {image_path} or could not be loaded.")

    # Display the grayscale image
    plt.imshow(gray_image, cmap='gray')
    plt.title('Grayscale Image')
    plt.axis('off') # Hide axes ticks
    plt.show()

except Exception as e:
    print(f"An error occurred: {e}")
    print("Creating a simple black and white gradient image for demonstration...")
    # Create a simple grayscale gradient image if reading fails
    height, width = 200, 300
    gradient_image = np.zeros((height, width), dtype=np.uint8)
    for i in range(width):
        gradient_image[:, i] = int(255 * (i / (width - 1)))

    plt.imshow(gradient_image, cmap='gray')
    plt.title('Generated Grayscale Gradient Image')
    plt.axis('off')
    plt.show()
    print("Please replace 'sample_image.jpg' with the actual path to your image file for a real test.")
```

### Explanation:
1.  **Install Libraries**: We install `opencv-python` for image processing and `matplotlib` for displaying images.
2.  **Import Libraries**: `cv2` is used for reading and processing images, `matplotlib.pyplot` for plotting.
3.  **Image Loading**: `cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)` reads the image specified by `image_path`. The `cv2.IMREAD_GRAYSCALE` flag ensures that the image is loaded directly as a single-channel grayscale image.
    *   A small utility is included to download a sample image if `sample_image.jpg` is not found, making the code runnable out-of-the-box.
    *   Error handling is included to create a simple gradient image if loading fails.
4.  **Display Image**: `plt.imshow(gray_image, cmap='gray')` displays the grayscale image. `cmap='gray'` ensures that `matplotlib` interprets and displays the single-channel image correctly as shades of gray. `plt.title()` adds a title, `plt.axis('off')` removes the axes, and `plt.show()` renders the plot.
