```python
import numpy as np

# Create a dummy image matrix (e.g., a grayscale image 10x10 pixels)
# The data type (dtype) is often uint8 for image pixel values
image_matrix = np.random.randint(0, 256, size=(10, 10), dtype=np.uint8)

# Print the data type of the image matrix
print(f"Data type of the image matrix: {image_matrix.dtype}")

# You can also create a color image matrix (e.g., 10x10 pixels with 3 color channels)
color_image_matrix = np.random.randint(0, 256, size=(10, 10, 3), dtype=np.uint8)
print(f"Data type of the color image matrix: {color_image_matrix.dtype}")
```

```python
from PIL import Image
import numpy as np

# Load the image
image_path = '/content/tigerr.jpg'
try:
    pil_image = Image.open(image_path)
    print(f"Image '{image_path}' loaded successfully.")
except FileNotFoundError:
    print(f"Error: Image '{image_path}' not found.")
    pil_image = None
```

```python
if pil_image:
    # Display the image
    display(pil_image)
```

```python
if pil_image:
    # Convert PIL Image to NumPy array to get the dtype
    image_array = np.array(pil_image)
    print(f"Data type of the loaded image array: {image_array.dtype}")
else:
    print("No image was loaded, cannot determine data type.")
```
