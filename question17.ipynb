```python
# Write a program to calculate and print the mean intensity of a grayscale image
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image in grayscale
image = cv2.imread("/content/lotus21.jpg", cv2.IMREAD_GRAYSCALE)

# Check whether image is loaded successfully
if image is None:
    print("Error: Image could not be loaded.")
else:
    # Calculate mean intensity
    mean_intensity = image.mean()

    # Print the mean intensity
    print("Mean intensity value:", mean_intensity)

    # Display the grayscale image
    plt.imshow(image, cmap="gray")
    plt.title(f"Grayscale Image\nMean Intensity: {mean_intensity:.2f}")
    plt.axis("off")
    plt.show()
```

# **Explanation**
cv2.imread() reads the image.
cv2.IMREAD_GRAYSCALE loads the image as a grayscale image.
image.mean() calculates the average intensity of all pixels.
print() displays the mean intensity value in the terminal.
plt.imshow() displays the grayscale image.
cmap="gray" ensures the image is displayed in grayscale.
plt.title() displays the mean intensity along with the image.
