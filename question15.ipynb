```python
# Write a program to merge three separate image channels into a single color image.
```

```python
import numpy as np
from PIL import Image

# Define the dimensions of the image
width, height = 256, 256

# 1. Create three separate 8-bit grayscale NumPy arrays to represent R, G, B channels
# Red channel: gradient from black to red horizontally
red_channel_np = np.zeros((height, width), dtype=np.uint8)
for x in range(width):
    red_channel_np[:, x] = int(x / width * 255)

# Green channel: gradient from black to green vertically
green_channel_np = np.zeros((height, width), dtype=np.uint8)
for y in range(height):
    green_channel_np[y, :] = int(y / height * 255)

# Blue channel: solid blue (or inverse gradient for distinction)
blue_channel_np = np.full((height, width), 255 - np.arange(width) * (255 // width), dtype=np.uint8)

print("Created three NumPy arrays for R, G, B channels.")
```

```python
# 2. Convert NumPy arrays to PIL Image objects (mode 'L' for grayscale)
red_image = Image.fromarray(red_channel_np, mode='L')
green_image = Image.fromarray(green_channel_np, mode='L')
blue_image = Image.fromarray(blue_channel_np, mode='L')

print("Converted NumPy arrays to PIL Image objects.")
```

```python
# 3. Merge the three grayscale image channels into a single RGB color image
merged_image = Image.merge('RGB', (red_image, green_image, blue_image))

print("Merged channels into a single RGB image.")

# 4. Display the resulting color image
print("Displaying the merged image:")
display(merged_image)
```

Creating NumPy Arrays for Channels: We first use the numpy library to create three separate 2D arrays, red_channel_np, green_channel_np, and blue_channel_np. These arrays are initialized with np.zeros or np.full and are of uint8 data type, meaning each pixel value will be an integer between 0 and 255. Each array represents a single color channel (Red, Green, Blue) as a grayscale image.

The red channel has a horizontal gradient, making areas redder as you move from left to right.
The green channel has a vertical gradient, making areas greener as you move from top to bottom.
The blue channel has an inverse horizontal gradient.
Converting to PIL Image Objects: The Pillow library (PIL) is then used to convert these numpy arrays into Image objects. Image.fromarray(array, mode='L') creates a grayscale PIL image from each numpy array. The 'L' mode signifies that it's a luminance (grayscale) image.

Merging Channels into an RGB Image: The core of the merging process happens with Image.merge('RGB', (red_image, green_image, blue_image)). This function takes a mode string (here, 'RGB') and a tuple of individual channel images. It combines the three grayscale PIL images (one for red, one for green, one for blue) into a single full-color RGB PIL image.

Displaying the Result: Finally, display(merged_image) is used to show the resulting colorful image directly in the notebook output.
