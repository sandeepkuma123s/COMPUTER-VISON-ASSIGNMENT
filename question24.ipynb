```python
# Write a program to crop a rectangular Region of Interest (ROI) from an image using user-provided coordinates.
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image
image = cv2.imread("/content/dog.webp")

if image is None:
    print("Error: Image could not be loaded.")
else:
    height, width = image.shape[:2]

    print("Image width:", width)
    print("Image height:", height)

    # User-provided coordinates
    x1 = int(input("Enter x1: "))
    y1 = int(input("Enter y1: "))
    x2 = int(input("Enter x2: "))
    y2 = int(input("Enter y2: "))

    # Check coordinates
    if x1 < 0 or y1 < 0 or x2 > width or y2 > height:
        print("Error: Coordinates are outside the image.")
    elif x1 >= x2 or y1 >= y2:
        print("Error: Invalid coordinates.")
    else:
        # Crop ROI
        roi = image[y1:y2, x1:x2]

        # Display
        plt.figure(figsize=(10, 5))

        plt.subplot(1, 2, 1)
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
        plt.title("Original Dog Image")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
        plt.title("Cropped ROI")
        plt.axis("off")

        plt.show()
```

# Explanation

The program is used to crop a rectangular Region of Interest (ROI) from an image.

First, the image is loaded and its width and height are obtained.
The user enters four coordinates: x1, y1, x2, y2.
(x1, y1) represents the top-left corner of the ROI.
(x2, y2) represents the bottom-right corner.
The selected rectangular area is cropped and displayed along with the original image.
The coordinates must be within the image dimensions.

For our dog image: x1=50, y1=20, x2=225, y2=220.

# Conclusion

The program successfully extracts and displays the required rectangular ROI from the original image using user-provided coordinates.
