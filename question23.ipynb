```python
# Write a program to downsample an image by a factor of 2 in both width and height and print the original and new resolutions.
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image
image = cv2.imread("/content/lotus2.jpg")

if image is None:
    print("Error: Image could not be loaded.")
else:
    # Original resolution
    height, width = image.shape[:2]

    # Downsample by factor of 2
    downsampled = cv2.resize(image, (width // 2, height // 2))

    # New resolution
    new_height, new_width = downsampled.shape[:2]

    print("Original resolution:", width, "x", height)
    print("New resolution:", new_width, "x", new_height)

    # Display images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(downsampled, cv2.COLOR_BGR2RGB))
    plt.title("Downsampled Image")
    plt.axis("off")

    plt.show()
```

# Explanation
The original image is loaded using OpenCV.
Its width and height are divided by 2.
cv2.resize() creates the smaller image.
The program prints both the original and new resolutions.
# Conclusion

The image is successfully downsampled to half its original width and half its original height.
